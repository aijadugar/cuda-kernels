```python
thread_idx=r"""

#include <cstdio>
#include <cuda_runtime.h>

__global__ void wai(){
    int x=blockIdx.x*blockDim.x+threadIdx.x;
    int y=blockIdx.y*blockDim.y+threadIdx.y;

    printf("block(%d, %d) thread(%d, %d) -> global(x=%d, y=%d)\n", blockIdx.x, blockIdx.y, threadIdx.x, threadIdx.y, x, y);
}

int main(){
    dim3 grid(2, 2);
    dim3 block(3, 2);

    wai<<<grid, block>>>();
    cudaDeviceSynchronize();

return 0;
}

"""

with open('thread_idx.cu', 'w') as f:
    f.write(thread_idx)
```
