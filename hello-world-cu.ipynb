```python
hello = r"""
#include <iostream>
#include <cstdio>

__global__ void cuda_hello(){
    printf("Hello world from GPU thread %d!\n", threadIdx.x);
}

int main(){
    std::cout<< "Hello world from CPU!" << std::endl;

    cuda_hello<<<1, 5>>>();
    cudaDeviceSynchronize();
    fflush(stdout);

return 0;
}
"""

with open("main.cu", 'w') as f:
    f.write(hello)
```

```python
!ls
```

```python
!nvcc -arch=sm_75 main.cu -o main
```
